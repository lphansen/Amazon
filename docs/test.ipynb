```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Example DataFrames for Model A and Model B
df_A = pd.DataFrame({
    'Parameter': ['Alpha', 'Beta', 'Gamma'],
    'Value': [0.5, 0.25, 0.75]
})

df_B = pd.DataFrame({
    'Parameter': ['Alpha', 'Beta', 'Gamma'],
    'Value': [0.15, 0.35, 0.55]
})

# Widget to select the model
model_selector = widgets.Dropdown(
    options=['Model A', 'Model B'],
    value='Model A',
    description='Select Model:',
    style={'description_width': 'initial'}
)

# Output widget to display tables
output = widgets.Output()

def on_model_change(change):
    model = change['new']
    if model == 'Model A':
        data = df_A
    else:
        data = df_B

    with output:
        output.clear_output()
        display(data)

model_selector.observe(on_model_change, names='value')

# Display the widget and output initially
display(model_selector, output)
on_model_change({'new': model_selector.value})  # Initial call to display data
```

```python
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, HTML
from IPython.display import display, Math, Latex

def rename(df):
    df.columns=['Pa','Pe','b','agricultural output','net transfers','forest services','adjustment costs','planner value']
    # Define the custom values for the first row
    custom_values = ['($)'] + ['($)'] + ['($)'] + ['($10^11)'] * (len(df.columns) - 3)
    
    # Create a new DataFrame for the new row with the custom values
    new_row = pd.DataFrame([custom_values], columns=df.columns)
    
    # Concatenate this new row to the top of the existing DataFrame
    df = pd.concat([new_row, df], ignore_index=True)
    
    return df
# Example DataFrames for Model A and Model B
df_det_1043 = pd.read_csv('data/1043site/det/pv_41.11.csv')
df_det_1043 = rename(df_det_1043)
# df_det_78_41 = pd.read_csv('data/78site/det/pv_41.11.csv')
# df_det_78_35 = pd.read_csv('data/78site/det/pv_35.71.csv')
# df_det_78_44 = pd.read_csv('data/78site/det/pv_44.26.csv')
# df_det_78 = pd.concat([df_det_78_35, df_det_78_41, df_det_78_44], ignore_index=True)
df_det_78 = pd.read_csv('data/78site/det/pv_41.11.csv')
df_det_78 = rename(df_det_78)
df_xi1 = pd.read_csv('data/78site/hmc/pv_xi1.csv')
df_xi1 = rename(df_xi1)
df_xi2 = pd.read_csv('data/78site/hmc/pv_xi2.csv')
df_xi2 = rename(df_xi2)
df_mpc=pd.read_csv('data/78site/mpc/mpc.csv',na_filter=False)


def create_df_widget(df,subtitle):
    """Utility function to create a widget for displaying a DataFrame with centered cells."""
    # Define CSS to center text in table cells
    style = """
    <style>
        .dataframe td, .dataframe th {
            text-align: center;
            vertical-align: middle;
        }
        .dataframe thead th {
            background-color: #f2f2f2;  # Light gray background in the header
        }
    </style>
    """
    html = style + df.to_html(index=False)
    html_widget = widgets.HTML(value=html)  # Use ipywidgets.HTML here
    subtitle_widget = widgets.Label(value=subtitle, layout=widgets.Layout(justify_content='center'))
    out = widgets.VBox([subtitle_widget, html_widget], layout={'border': '1px solid black'})
    return out




# Tab widget to hold different models
tab = widgets.Tab()
children = [create_df_widget(df_det_1043,'Present-value decomposition - 1043 sites'),
            create_df_widget(df_det_78,'Present-value decomposition - 78 sites'), 
            create_df_widget(df_xi2,'Present-value decomposition - parameter ambiguity'),
            create_df_widget(df_xi1,'Present-value decomposition - parameter ambiguity'),
            create_df_widget(df_mpc,'Present-value decomposition with stochastic agricultural prices')]
tab.children = children
for i, title in enumerate(['1043site ξ = ∞ ','78site ξ = ∞ ', '78site ξ = 2','78site ξ = 1','78site Pa stochastic']):
    tab.set_title(i, title)

# Display the tab widget
tab.selected_index = 1
display(tab)
```
